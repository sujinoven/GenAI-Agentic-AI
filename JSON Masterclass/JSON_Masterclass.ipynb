# Level 1 — JSON Fundamentals

## 1.1 What is JSON?

JSON (JavaScript Object Notation) is a structured text format commonly used to exchange data between applications.

The basic pattern is:

```text
{
    "KEY": VALUE
}
```

The most important JSON value types are:
- **String** → text
- **Number** → numeric value
- **Boolean** → `true` or `false`
- **Null** → no value
- **Object** → `{}`
- **Array** → `[]`

````python
customer = {
    "name": "Sujin",
    "age": 25,
    "city": "Chennai",
    "is_active": True,
    "middle_name": None,
}

print(customer)
````

### JSON version of the same structure

Notice the JSON rules:
- Keys use double quotes.
- Strings use double quotes.
- JSON booleans are lowercase: `true` / `false`.
- JSON null is lowercase: `null`.
- Items are separated by commas.

````python
{
    "name": "Sujin",
    "age": 25,
    "city": "Chennai",
    "is_active": True,
    "middle_name": None
}
````

# Level 2 — Objects, Arrays, and Nested JSON

## 2.1 Object

An object is a collection of key-value pairs.

```json
{
    "name": "John",
    "city": "Bangalore"
}
```

## 2.2 Array

An array stores multiple values.

```json
[
    "Python",
    "SQL",
    "Power BI"
]
```

Remember:
- `{}` → Object
- `[]` → Array

````python
customer = {
    "name": "John",
    "email": "john@example.com",
    "orders": [
        "ORD-101",
        "ORD-102",
        "ORD-103",
    ],
}

print(customer)
````

## 2.3 Nested Objects

A JSON value can itself be another object.

````python
customer = {
    "customer": {
        "name": "John",
        "city": "Bangalore",
    }
}

print(customer)
````

### Think about the hierarchy

```text
Object
└── customer
    └── Object
        ├── name
        └── city
```

# Level 3 — JSON Schema

At Level 1 and Level 2, JSON represented **data**.

At Level 3, JSON starts describing **rules for data**.

The most important concepts are:
- `type`
- `properties`
- `required`
- `additionalProperties`

````python
customer_schema = {
    "type": "object",
    "properties": {
        "customer_id": {
            "type": "string",
        },
    },
    "required": [
        "customer_id",
    ],
    "additionalProperties": False,
}

print(customer_schema)
````

## Understanding each keyword

### `type`

```json
"type": "object"
```

The overall value must be an object.

### `properties`

```json
"properties": {
    "customer_id": {
        "type": "string"
    }
}
```

This defines the fields and their expected types.

### `required`

```json
"required": [
    "customer_id"
]
```

This says `customer_id` must be supplied.

### `additionalProperties`

```json
"additionalProperties": false
```

This prevents extra fields that were not defined in `properties`.

## Schema Mental Model

```text
SCHEMA
│
├── type
│
├── properties
│     └── customer_id → string
│
├── required
│     └── customer_id
│
└── additionalProperties
      └── false
```

# Level 4 — Function / Tool Definition

Now we use JSON to describe a function that an AI model can call.

The basic structure is:

```text
Function
├── type
├── name
├── description
└── parameters
```

````python
get_customer_profile_tool = {
    "type":"function",
    "name":"get_customer_prodile",
    "description":"Fetch a Customer profile by Customer ID.",
    "parameters":{
        "type":"object",
        "properties":{
            "customer_id":{
                "type":"string",
            },
        },
        "required":["customer_id"],
        "additional_properties":False
    },
}
print(get_customer_profile_tool)
````

## What does `parameters` mean?

The `parameters` section does **not** contain the actual customer ID.

It describes the input that the function expects.

Function:

```text
get_customer_profile()
```

Expected input:

```json
{
    "customer_id": "CUST-12345"
}
```

# Level 5 — Namespace + Multiple Tools

A **namespace** is a logical container used to group related tools or capabilities.

Think of it like a folder:

```text
CRM Namespace
├── get_customer_profile()
├── list_open_orders()
└── update_customer()
```

The key distinction is:

> **Object groups related data. Namespace groups related tools/capabilities.**

````python
crm_namespace = {
    "type": "namespace",
    "name": "crm",
    "description": "CRM tools for customer lookup and order management.",
    "tools": [
        {
            "type": "function",
            "name": "get_customer_profile",
            "description": "Fetch a customer profile by customer ID.",
            "parameters": {
                "type": "object",
                "properties": {
                    "customer_id": {
                        "type": "string",
                    },
                },
                "required": [
                    "customer_id",
                ],
                "additionalProperties": False,
            },
        },
        {
            "type": "function",
            "name": "list_open_orders",
            "description": "List open orders for a customer ID.",
            "defer_loading": True,
            "parameters": {
                "type": "object",
                "properties": {
                    "customer_id": {
                        "type": "string",
                    },
                },
                "required": [
                    "customer_id",
                ],
                "additionalProperties": False,
            },
        },
    ],
}

print(crm_namespace)
````

## Read the namespace from outside → inside

```text
namespace
   ↓
crm
   ↓
tools
   ↓
function
   ↓
name + description
   ↓
parameters
   ↓
object
   ↓
properties
   ↓
customer_id
   ↓
string
```

## Understanding `tools: [...]`

`tools` is an array because a namespace can contain multiple tools.

```text
CRM
│
└── tools [ ]
      │
      ├── Tool 1
      │     └── get_customer_profile()
      │
      └── Tool 2
            └── list_open_orders()
```

## Understanding `defer_loading`

```json
"defer_loading": true
```

This configures the tool for deferred loading. In an agent workflow, the tool can be discovered when it is needed rather than being part of the initially loaded tool definitions.

Conceptually:

```text
AI Agent
   │
   ├── Available tools
   │
   └── Tool Search
          │
          └── Discover deferred tool
                 │
                 └── list_open_orders()
```

# Final Practice — Task 5

### Build an AI tool namespace from a business scenario

Imagine you are building a customer-support AI agent for a company. The agent needs a `support` namespace containing two tools:

1. A tool called `get_ticket` that fetches a support ticket using a required `ticket_id` string.
2. A tool called `close_ticket` that closes a support ticket using a required `ticket_id` string.

**Task:** Convert this requirement into a complete namespace structure using JSON/Python dictionary syntax.

Make sure you correctly use:
- `namespace`
- `tools` as an array
- `function`
- `name`
- `description`
- `parameters`
- `properties`
- `required`
- `additionalProperties`

````python
# Solution

support_namespace = {
    "type": "namespace",
    "name": "support",
    "description": "Support tools for ticket lookup and ticket management.",
    "tools": [
        {
            "type": "function",
            "name": "get_ticket",
            "description": "Fetch a support ticket by ticket ID.",
            "parameters": {
                "type": "object",
                "properties": {
                    "ticket_id": {
                        "type": "string",
                    },
                },
                "required": [
                    "ticket_id",
                ],
                "additionalProperties": False,
            },
        },
        {
            "type": "function",
            "name": "close_ticket",
            "description": "Close a support ticket by ticket ID.",
            "parameters": {
                "type": "object",
                "properties": {
                    "ticket_id": {
                        "type": "string",
                    },
                },
                "required": [
                    "ticket_id",
                ],
                "additionalProperties": False,
            },
        },
    ],
}

print(support_namespace)
````

# Quick Cheat Sheet

| Concept | Meaning |
|---|---|
| `{}` | Object |
| `[]` | Array / list |
| `"type"` | Defines the expected type |
| `"properties"` | Defines available fields |
| `"required"` | Defines mandatory fields |
| `"additionalProperties": false` | Rejects undefined extra fields |
| `"type": "function"` | Describes a callable tool/function |
| `"name"` | Function/tool name |
| `"description"` | Explains what the tool does |
| `"parameters"` | Defines the tool's expected input |
| `namespace` | Groups related tools/capabilities |
| `tools` | Array of tools inside a namespace |
| `defer_loading` | Configures deferred tool loading |

## One sentence to remember

> **Object groups data. Array groups items. Schema defines rules. Function schema describes a tool. Namespace groups related tools.**
