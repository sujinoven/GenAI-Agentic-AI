```python
from openai import OpenAI

client = OpenAI(api_key="xyzxyszxysz")

response = client.responses.create(model="gpt-4o-mini",
                                  input=[{"role":"user","content":"Knock Knock."},
                                          {"role":"assistant","content":"Who's there"},
                                          {"role":"user","content":"Orange."}])

print(response.output_text)
```

## The Right Strategy - To Add the conversation Memory

```python
from openai import OpenAI

client = OpenAI(api_key="xyzxysxyz")

conversation_history = []

while True:
    question = input("Ask something: ")
    if question.lower() == "exit":
        break

    conversation_history.append({"role": "user", "content": question})

    response = client.chat.completions.create(model="gpt-5.6",
                                              messages= conversation_history)
    answer = response.choices[0].message.content
    conversation_history.append({"role": "assistant", "content": answer})
    print("AI answer is:",answer)
```

## Bottleneck

Imagine a Conversation with 1000 messages.

Should we continue the entire conversation every time?

No,And that's the bottleneck.
