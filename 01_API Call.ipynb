### 1.GET request

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/todos/1")

print("Status Code:",response.status_code)
print(response.text)
```

### 2.Understanding JSON

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/todos/2")

data = response.json()

print(data)
print(type(data))
print(data["title"])
```

## JSON is the standard format,applications use to exchange structured information.

## 3.POST request

URL:https:postman-echo.com/post

```python
import requests

details = {"name":"SUJIN",
          "Company":"Flyrank AI",
          "Role":"AI Engineer"}

response = requests.post("https://postman-echo.com/post",json=details,timeout=30)

print("Status Code:",response.status_code)
print(response.json())
```

## Takeaway:

1.GET retrieves information

2.POST sends structured information

3.Modern AI APIs internally use the same communication principles

## 5.Now understanding the Engineering problem

```python
from openai import OpenAI

client= OpenAI(api_key = "xyzxyzxyzxyzxyz")

question = input("Ask Anything:")

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
        "role":"user",
        "content":"question"
        }
    ]
)

print(response.choices[0].message.content)
```
